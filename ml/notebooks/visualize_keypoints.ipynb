# Dataset Keypoints Visualization
This notebook visualizes the keypoints and skeletons overlaid on sample images from the COCO and MPII datasets.

```python
import sys
import os
import matplotlib.pyplot as plt
import numpy as np

sys.path.append(os.path.abspath('../src'))
from loaders import COCOLoader, MPIILoader

# Skeleton definitions (pairs of keypoint indices)
COCO_SKELETON = [[15,13], [13,11], [16,14], [14,12], [11,12], [5,11], [6,12], [5,6], [5,7], [6,8], [7,9], [8,10], [1,2], [0,1], [0,2], [1,3], [2,4], [3,5], [4,6]]
MPII_SKELETON = [[0,1], [1,2], [2,6], [6,3], [3,4], [4,5], [6,7], [7,8], [8,9], [10,11], [11,12], [12,7], [13,7], [13,14], [14,15]]

def draw_skeleton(ax, keypoints, skeleton, threshold=0.1):
    if len(keypoints) == 0:
        return
    
    for kpts in keypoints:
        # Scatter points
        for i in range(len(kpts)):
            if len(kpts[i]) > 2 and kpts[i, 2] > threshold:
                ax.plot(kpts[i, 0], kpts[i, 1], 'ro', markersize=4)
        
        # Draw lines
        for pair in skeleton:
            pt1, pt2 = pair
            if pt1 < len(kpts) and pt2 < len(kpts):
                if (len(kpts[pt1]) <= 2 or kpts[pt1, 2] > threshold) and \
                   (len(kpts[pt2]) <= 2 or kpts[pt2, 2] > threshold):
                    ax.plot([kpts[pt1, 0], kpts[pt2, 0]], 
                            [kpts[pt1, 1], kpts[pt2, 1]], 'b-', linewidth=2)
```

## COCO Dataset
Assuming annotations exist at `../datasets/coco/annotations/person_keypoints_train2017.json`

```python
coco_ann = '../datasets/coco/annotations/person_keypoints_train2017.json'
coco_img_dir = '../datasets/coco/train2017'

if os.path.exists(coco_ann):
    coco_loader = COCOLoader(coco_ann, coco_img_dir)
    fig, axes = plt.subplots(1, 5, figsize=(20, 5))
    for i in range(5):
        try:
            img, kpts = coco_loader.get_sample(i)
            axes[i].imshow(img)
            draw_skeleton(axes[i], kpts, COCO_SKELETON)
            axes[i].axis('off')
        except Exception as e:
            axes[i].set_title("Error loading")
            print(f"Error on COCO {i}: {e}")
    plt.show()
else:
    print("COCO annotations not found. Please follow README instructions to download.")
```

## MPII Dataset
Assuming annotations exist at `../datasets/mpii/mpii_annotations.json`

```python
mpii_ann = '../datasets/mpii/mpii_annotations.json'
mpii_img_dir = '../datasets/mpii/images'

if os.path.exists(mpii_ann):
    mpii_loader = MPIILoader(mpii_ann, mpii_img_dir)
    fig, axes = plt.subplots(1, 5, figsize=(20, 5))
    for i in range(5):
        try:
            img, kpts = mpii_loader.get_sample(i)
            axes[i].imshow(img)
            draw_skeleton(axes[i], kpts, MPII_SKELETON)
            axes[i].axis('off')
        except Exception as e:
            axes[i].set_title("Error loading")
            print(f"Error on MPII {i}: {e}")
    plt.show()
else:
    print("MPII annotations not found. Please follow README instructions to download.")
```
